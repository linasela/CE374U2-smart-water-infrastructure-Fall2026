### A simple engineering calculation

Suppose a pump delivers 150 gallons per minute for 4 hours. We can calculate the total delivered volume.

````python
# Pump delivery example

flow_rate_gpm = 150      # gallons per minute
duration_hours = 4       # hours

volume_gallons = flow_rate_gpm * 60 * duration_hours

print(f'Total delivered volume = {volume_gallons} gallons')
````

**What just happened?**

1. We stored known values using variable names.
2. We performed a calculation.
3. We displayed the result.

That pattern appears throughout engineering analysis:

**Inputs → Calculation → Output**

### Concept 3: Collections of Data

Many engineering analyses involve groups of values rather than individual numbers.

Python provides several ways to store collections of data. We will start with **lists**, and later learn about **NumPy arrays** and **Pandas DataFrames** for larger datasets.

A list stores multiple values in order.

💡 **Indexing**

Each value in a list has a position called an **index**.

Python uses **zero-based indexing**, meaning the first element has index `0`.

For the list below:

| Index | Value |
|---------|---------|
| 0 | 2.1 |
| 1 | 2.4 |
| 2 | 3.2 |
| 3 | 4.6 |
| 4 | 5.1 |
| 5 | 4.3 |
| 6 | 3.0 |

This allows us to access individual values using square brackets.

````python
hourly_demand_mgd = [2.1, 2.4, 3.2, 4.6, 5.1, 4.3, 3.0] # this is a list

print(f'First value: {hourly_demand_mgd[0]} MGD')   # Index 0 = first element
print(f'Third value: {hourly_demand_mgd[2]} MGD')   # Index 2 = third element
print(f'Last value: {hourly_demand_mgd[-1]} MGD')   # Index -1 = last element

print(f'Number of values: {len(hourly_demand_mgd)}')  # using len function
print(f'Maximum demand: {max(hourly_demand_mgd)} MGD')  # using max function
````

## <font color="blue"><b>PART 5: YOUR FIRST SMART WATER SYSTEMS ANALYSIS</b></font>

Now we will analyze a simple water demand time series.

This example is small, but it uses the same structure we will use later with real datasets:

1. Define data
2. Calculate summary statistics
3. Check for unusual values
4. Create a plot
5. Interpret the results

````python
# Hourly demand data for one day
hours = [0, 3, 6, 9, 12, 15, 18, 21, 24]
demand_mgd = [2.1, 1.8, 2.6, 4.2, 5.0, 4.6, 5.4, 3.8, 2.4]
````

````python
# Summary statistics
peak_demand = max(demand_mgd)
average_demand = sum(demand_mgd) / len(demand_mgd)
````

**Note:** `max()`, `sum()`, and `len()` are built-in Python functions, so they can be used without importing any additional libraries.

````python
print('DAILY DEMAND SUMMARY')
print('=' * 30)
print(f'Peak demand: {peak_demand:.1f}')
print(f'Average demand: {average_demand:.1f} MGD')
````

Create a line plot with:

- `hours` on the x-axis;
- `demand_mgd` on the y-axis;
- circular markers;
- an x-axis label;
- a y-axis label with units;
- a descriptive title; and
- grid lines.


````python
# Create a plot
# we need to import a package that can make plots; we'll talk about this more later in the class
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))                                # create figure
plt.plot(hours, demand_mgd, marker='o', linewidth=2)      # plot(x,y,optional input)
plt.xlabel('Hour of day')                                 # add x label
plt.ylabel('Demand (MGD)')                                # add y label
plt.title('Example Daily Water Demand Pattern')           # add title
plt.grid(True)                                            # add grid
plt.show()
````

### Interpreting the plot

Think like an engineer:

1. When is demand highest?
2. When is demand lowest?
3. Does the pattern make sense for a daily water demand curve?
4. What additional information would you want before making an operational decision?

### 🎯Add a simple operational check

Suppose the system has an operational threshold of 5.0 MGD. We can check which time periods exceed that threshold.

````python
threshold_mgd = 5.0

print('DEMAND THRESHOLD CHECK')
print('=' * 30)

peak_demand = max(demand_mgd)

if peak_demand > threshold_mgd:
    print('Peak demand exceeds threshold.')
else:
    print('Peak demand is below threshold.')
````

### 🎯Exercise 3: Explore pressure measurements
The list below contains eight pressure measurements in pounds per square inch (psi).

Run the cell, then add code that calculates and prints:

1. the number of measurements;
2. the minimum pressure; and
3. the average pressure, rounded to one decimal place.


````python
# Part 3: Pressure measurements

pressure_values = [54, 50, 46, 38, 41, 49, 52, 47]

number_of_measurements = len(pressure_values)
minimum_pressure = min(pressure_values)
average_pressure = sum(pressure_values) / len(pressure_values)

print(f'Number of measurements: {number_of_measurements}')
print(f'Minimum pressure: {minimum_pressure} psi')
print(f'Average pressure: {average_pressure:.1f} psi')
````

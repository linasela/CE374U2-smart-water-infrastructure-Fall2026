## <font color="blue"><b>PART 3: IMPORTING CSV DATA</b></font>

A CSV file is a plain text file that stores data in rows and columns.

CSV stands for **comma-separated values**.

Before importing a CSV file into Python, it is useful to open it and look at the structure:

- How many rows are there?
- How many columns are there?
- Are there column names?
- Are values separated by commas?
- Are the values numeric?

In this example, we will load a CSV file into a NumPy array.

````python
# Import packages

import numpy as np
import matplotlib.pyplot as plt
````

💡 **Tip:** Import statements usually appear near the beginning of a notebook. They load tools we need for the analysis.

````python
# Define the path to the data file

data_file = os.path.join(data_folder, '01_02_working_with_files.csv')

print(f'Data file location:')
print(data_file)
````

````python
# Load the CSV file into a NumPy array

if os.path.exists(data_file):
    A = np.loadtxt(data_file, delimiter=',')
    print('Data loaded successfully.')
else:
    print('Data file not found.')
    print(f'Please upload 01_02_working_with_files.csv to:')
    print(data_folder)
````

The variable `A` stores the data from the CSV file.

In this notebook, `A` is a **NumPy array**, which is a structured collection of numeric values.

Later in the course, we will also use **Pandas DataFrames**, which are especially useful for datasets with column names, dates, and mixed data types.

## <font color="blue"><b>PART 4: UNDERSTANDING THE DATA STRUCTURE</b></font>

After importing data, one of the first things we should do is inspect its structure.

We can ask:

- What are the dimensions of the dataset?
- How many rows does it have?
- How many columns does it have?
- What do the first few rows look like?

````python
# Check the dimensions of the data

print(f'Dimensions of A: {A.shape}')
print(f'Number of rows: {A.shape[0]}')
print(f'Number of columns: {A.shape[1]}')
````

````python
# Display the first five rows

print('First five rows of A:')
print(A[:5, :])
````

### Understanding rows and columns

For a two-dimensional array:

```python
A[row, column]
```

Python indexing starts at 0.

That means:

- `A[0, 0]` is the value in the first row and first column
- `A[0, 1]` is the value in the first row and second column
- `A[:, 0]` selects all rows from the first column
- `A[:, 1]` selects all rows from the second column

````python
# Access a few values from the array

print(f'First row, first column: {A[0, 0]}')
print(f'First row, second column: {A[0, 1]}')
print(f'First column: {A[:, 0]}')
````

In this dataset, the columns are organized in pairs.

Each pair contains:

- one `x` column
- one `y` column

We will choose one pair of columns, calculate statistics, and create a plot.

## <font color="blue"><b>PART 5: BASIC STATISTICS</b></font>

We will choose a data pair and compute summary statistics.

Summary statistics are helpful, but they do not tell the whole story. Two datasets can have similar means and standard deviations but very different patterns.

That is why we will also plot the data.

````python
# Choose a data pair

# Choose a number between 1 and 13
n = 1

# You can also choose randomly between 1 and 13 by uncommenting the line below
# n = np.random.randint(1, 14)

print(f'Selected data pair: {n}')
````

### Selecting columns

Because Python indexing starts at 0:

- data pair 1 uses columns 0 and 1
- data pair 2 uses columns 2 and 3
- data pair 3 uses columns 4 and 5

For data pair `n`:

- the x column is `2*n - 2`
- the y column is `2*n - 1`

````python
# Select the x and y columns for the chosen data pair

x_column = 2*n - 2
y_column = 2*n - 1

x = A[:, x_column]
y = A[:, y_column]

print(f'x column index: {x_column}')
print(f'y column index: {y_column}')
print(f'Number of x values: {len(x)}')
print(f'Number of y values: {len(y)}')
````

````python
# Compute basic statistics

x_mean = x.mean()
x_std = x.std()

y_mean = y.mean()
y_std = y.std()

print(f'Mean of x: {x_mean:.5f}')
print(f'Standard deviation of x: {x_std:.5f}')
print(f'Mean of y: {y_mean:.5f}')
print(f'Standard deviation of y: {y_std:.5f}')
````

💡 **Tip:** Printing intermediate results helps you inspect the values stored in your program. This is a common practice when exploring data and building models.

## <font color="blue"><b>PART 6: PLOTTING DATA</b></font>

Statistics summarize data using numbers.

Plots help us see patterns.

For engineering data, plotting is often one of the most important steps in understanding a dataset.

````python
# Plot the selected data pair

plt.figure(figsize=(6, 5))
plt.plot(x, y, 'ko')
plt.xlabel('x')
plt.ylabel('y')
plt.title(f'Data Pair {n}')
plt.grid(True)
plt.show()
````

### Save the figure

We can save the figure to the `figures` folder we created earlier.

````python
# Save the plot to the figures folder

figure_file = os.path.join(figures_folder, f'data_pair_{n}.png')

plt.figure(figsize=(6, 5))
plt.plot(x, y, 'ko')
plt.xlabel('x')
plt.ylabel('y')
plt.title(f'Data Pair {n}')
plt.grid(True)
plt.savefig(figure_file, dpi=300, bbox_inches='tight')
plt.show()

print(f'Figure saved to:')
print(figure_file)
````

````python
# Print statistics for all 13 data pairs

print('Pair   x mean    x std    y mean    y std')
print('-' * 43)

for pair_number in range(1, 14):
    x = A[:, 2*pair_number - 2]
    y = A[:, 2*pair_number - 1]

    print(
        f'{pair_number:>4}   '
        f'{x.mean():>6.2f}   '
        f'{x.std():>6.2f}   '
        f'{y.mean():>6.2f}   '
        f'{y.std():>6.2f}'
    )
    plt.figure(figsize=(6, 5))
    plt.plot(x, y, 'ko')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Data Pair {n}')
    plt.grid(True)
    plt.show()
````

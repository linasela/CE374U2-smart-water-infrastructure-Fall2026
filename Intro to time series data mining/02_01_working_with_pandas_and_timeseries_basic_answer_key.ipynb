## <font color="blue"><b>PART 1: FROM ARRAYS TO DATAFRAMES</b></font>

In the previous notebook, we used a **NumPy array**.

A NumPy array is useful when the dataset is mostly numerical and has a simple structure.

However, many engineering datasets are more complex. They may include:

- Dates and times
- Column names
- Sensor names
- Missing values
- Multiple variables
- Different data types

For these datasets, we often use a **Pandas DataFrame**.

A DataFrame is similar to a spreadsheet table. It has rows, columns, column names, and data values.

### Arrays vs DataFrames

| Tool | Best for | Example |
|---|---|---|
| NumPy array | Numerical data with a simple structure | matrix of x-y values |
| Pandas DataFrame | Tables with column names, dates, and mixed data types | sensor time series data |

### **Selecting values by position with `iloc`**

The method `iloc` selects rows and columns by their integer position.

For example:

```python
df.iloc[11, 1]
```

means:

- row position 11
- column position 1

Because Python indexing starts at 0, this corresponds to the 12th row and the 2nd column.

````python
# Select values by position

DMA1_hr12 = df.iloc[11, 1]
DMA2_hr20 = df.iloc[19, 2]

print(f'DMA1 value at row position 11: {DMA1_hr12}')
print(f'DMA2 value at row position 19: {DMA2_hr20}')
````

## <font color="blue"><b>PART 5: WORKING WITH TIME</b></font>

The `timestamp` column may look like dates and times, but Pandas may initially read it as text.

For time series analysis, we need to convert it into a datetime format.

A datetime object allows Python to understand calendar information such as:

- year
- month
- day
- hour
- day of week

Convert `df['timestamp']` to datetime using `pd.to_datetime()`

````python
# Check the data type before conversion

print(f'Data type before conversion: {df["timestamp"].dtype}')
````

💡 **Note:** Check the time format in the your data

````python
# Convert timestamp column to datetime

df['timestamp'] = pd.to_datetime(df['timestamp'], dayfirst=True)

print(f'Data type after conversion: {df["timestamp"].dtype}')
````

The option `dayfirst=True` tells Pandas that the date format places the day before the month.

For example:

```text
05/01/2026
```

is interpreted as January 5, 2026 if `dayfirst=True`.

Be careful with date formats, because different datasets may use different conventions.

````python
# Inspect the DataFrame after converting timestamps

df.info()
````

### **Create new columns from the timestamp**

Once the timestamp is stored as a datetime object, we can extract useful information from it.

For example, we can create new columns for:

- hour of day
- day of week
- date

````python
# Create new time-related columns

df['hour'] = df['timestamp'].dt.hour
df['day_of_week'] = df['timestamp'].dt.day_name()
df['date'] = df['timestamp'].dt.date

df[['timestamp', 'hour', 'day_of_week', 'date']].head()
````

💡 This is an important idea: we can create **new variables** from existing data.

In water systems, this helps us compare demand patterns by hour, day, week, or season.

## <font color="blue"><b>PART 6: TIME SERIES PLOTS</b></font>

A time series plot shows how a variable changes over time.

Here we will plot the DMA1 demand values against the timestamp.

````python
# Plot DMA1 as a time series

plt.figure(figsize=(10, 4))
plt.plot(df['timestamp'], df['DMA1'])
plt.gcf().autofmt_xdate()
plt.title('DMA 1 Demand Time Series')
plt.xlabel('Time')
plt.ylabel('Demand (L/s)')
plt.grid(True)
plt.show()
````

### **Selecting a date range**

The full time series provides an overview of the data.

In many analyses, we are interested in a specific time period, such as a day, a week, or a month. We can create a subset of the data by selecting a date range.

Lets select the seven-day period beginning at midnight on **January 15, 2022** and ending immediately before midnight on **January 22, 2022**.


````python
# Define start and end dates

start_date = '2021-01-15'
end_date = '2021-01-22'

# Create a filtered DataFrame containing only the selected dates
# The '&' symbol means BOTH conditions must be true:
# 1. timestamp is after the start date
# 2. timestamp is before the end date

df_subset = df[
    (df['timestamp'] >= start_date) &
    (df['timestamp'] <= end_date)
]

print(f'Number of records in subset: {len(df_subset)}')
````

💡 How many data records you expect in an hourly dataset covering seven days?


Using `<= end_date` **includes** January 22 and gives exactly seven complete days.
This is actually a very common and useful convention for time-series filtering: **include the start, exclude the end**.

````python
# let's try again

df_subset = df[
    (df['timestamp'] >= start_date) &
    (df['timestamp'] < end_date)
]

print(f'Number of records in subset: {len(df_subset)}')
````

**Calculate** and print the mean and maximum of `DMA1` to two decimal places.

````python
mean_dma1 = df_subset['DMA1'].mean()
maximum_dma1 = df_subset['DMA1'].max()

print(f'Mean DMA2 demand: {mean_dma1:.2f} L/s')
print(f'Maximum DMA2 demand: {maximum_dma1:.2f} L/s')
````

**Compare** with the mean and maximum for the entire data

````python
mean_dma1 = df['DMA1'].mean()
maximum_dma1 = df['DMA1'].max()

print(f'Mean DMA2 demand: {mean_dma1:.2f} L/s')
print(f'Maximum DMA2 demand: {maximum_dma1:.2f} L/s')
````

````python
# Plot a selected time period

plt.figure(figsize=(10, 4))
plt.plot(df_subset['timestamp'], df_subset['DMA1'])

plt.gcf().autofmt_xdate()
plt.title('DMA 1 Demand (Selected Period)')
plt.xlabel('Time')
plt.ylabel('Demand (L/s)')
plt.grid(True)

plt.show()
````

Time series plots show how values change over time.

### **Histogram**

A histogram shows how often different values occur. This helps us see the distribution of demand values.

````python
# Histogram of DMA1 values

plt.figure(figsize=(7, 4))
plt.hist(df['DMA1'], bins=30, edgecolor='black')
plt.title('Histogram of DMA 1 Demand')
plt.xlabel('Demand (L/s)')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
````

### **Boxplot by day of week**

A box plot provides a simple way to compare the distribution of values across different groups. Here, we use a box plot to compare the distribution of water demand for each **day of the week**.

For each day:

- The **line inside the box** is the median, or middle value.
- The **box** contains the middle 50% of the data.
- The **whiskers** extend to values outside the box that are still within the typical range of the data.
- Individual points beyond the whiskers are considered **potential outliers**.

Box plots are useful for quickly comparing the typical demand, variability, and unusual values across different days of the week.

Here, we compare DMA1 demand by day of week.

### 💡Ordering categorical data

By default, Python does not know the natural order of the days of the week.  
`pd.Categorical()` tells pandas that `day_of_week` contains a specific set of categories and defines the order of those categories.

Here, we specify the order as Monday through Sunday and use `ordered=True` to tell pandas that this order is meaningful.

This is useful later when sorting, grouping, or plotting the data so that the days appear in calendar order rather than alphabetical order.

````python
# Create an ordered day-of-week column

day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

df['day_of_week'] = pd.Categorical(
    df['day_of_week'],
    categories=day_order,
    ordered=True
)

df[['timestamp', 'DMA1', 'day_of_week']].head()
````

````python
df.head()
````

### Creating a box plot

The `boxplot()` function creates a box plot from a DataFrame.

- `column='DMA1'` specifies the numerical data that we want to plot.
- `by='day_of_week'` specifies how we want to group the data. A separate box is created for each day of the week.
- `figsize=(12, 5)` sets the width and height of the figure.

So, this command shows the distribution of **DMA1 demand for each day of the week**.

````python
# Boxplot of DMA1 by day of week

df.boxplot(column='DMA1', by='day_of_week', figsize=(12, 5))
plt.title('DMA 1 Demand by Day of Week')
plt.suptitle('')
plt.xlabel('Day of week')
plt.ylabel('Demand (L/s)')
plt.grid(True)
plt.show()
````

## <font color="blue"><b>PART 8: MISSING DATA CHECKS</b></font>

Real sensor datasets often include missing values.

Missing values may occur because of:

- sensor failures
- communication problems
- data logging issues
- maintenance activities

Before analyzing a dataset, always check for missing values.

````python
# Count missing values in each column

missing_values = df.isnull().sum()

print('Missing values by column:')
print(missing_values)
````

````python
# Count missing values in DMA1 and DMA2

missing_dma1 = df['DMA1'].isnull().sum()
missing_dma2 = df['DMA2'].isnull().sum()

print(f'Number of missing values in DMA1: {missing_dma1}')
print(f'Number of missing values in DMA2: {missing_dma2}')
````
